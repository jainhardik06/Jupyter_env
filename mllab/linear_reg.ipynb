````python
# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Create Dataset
data = {
    'Experience': [1, 2, 3, 4, 5],
    'Salary': [30000, 35000, 40000, 45000, 50000]
}

df = pd.DataFrame(data)

# Step 3: Split Data
X = df[['Experience']]
y = df['Salary']

# Step 4: Train Model
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

# Step 5: Prediction
pred = model.predict(X)

print("Predicted Values:\n", pred)

# Step 6: Visualization
plt.scatter(X, y, color='blue')     # actual data
plt.plot(X, pred, color='red')      # regression line
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Linear Regression")
plt.show()
````

# 🧪 Experiment 2: Linear Regression

## 🔹 What is Linear Regression?

Linear Regression is a **supervised machine learning algorithm** used to predict a **continuous value**.

👉 Example:

* Predict salary based on experience
* Predict marks based on study hours

---

## 🔹 Core Concept

It finds a **best-fit straight line** between input (X) and output (Y).

* **y** → predicted value
* **x** → input feature
* **m** → slope (relationship strength)
* **c** → intercept

👉 Goal: minimize error between actual and predicted values.

# 🧠 Concepts You Must Understand

---

## 🔸 1. Supervised Learning

* Model learns from **labeled data**
* Input + Output already given

---

## 🔸 2. Best Fit Line

* Line that minimizes error
* Uses **least squares method**

---

## 🔸 3. Error (Loss)

* Difference between actual & predicted value

👉 Goal: minimize error

---

## 🔸 4. R² Score (Accuracy Measure)

* Value between 0 and 1
* Closer to 1 → better model

# ❓ Viva Questions with Answers

---

## ✅ Q1. What is Linear Regression?

Linear Regression is a supervised learning algorithm used to predict continuous values by finding a linear relationship between input and output variables.

---

## ✅ Q2. What type of problem does it solve?

It solves **regression problems** (continuous output).

---

## ✅ Q3. What is the equation of Linear Regression?

Already shown above in the widget — it represents a straight-line relationship between input and output.

---

## ✅ Q4. What is slope (m)?

Slope represents the rate of change of output with respect to input.

---

## ✅ Q5. What is intercept (c)?

Intercept is the value of y when x = 0.

---

## ✅ Q6. What is the goal of Linear Regression?

To find the best-fit line that minimizes the error between predicted and actual values.

---

## ✅ Q7. What is error in Linear Regression?

Error is the difference between actual value and predicted value.

---

## ✅ Q8. What is R² score?

R² score measures how well the model fits the data.
Value lies between 0 and 1.

---

## ✅ Q9. Difference between Regression and Classification?

| Regression        | Classification     |
| ----------------- | ------------------ |
| Continuous output | Discrete output    |
| Example: Salary   | Example: Pass/Fail |

---

## ✅ Q10. What are assumptions of Linear Regression?

* Linear relationship
* No multicollinearity
* Homoscedasticity (constant variance)
* No outliers
