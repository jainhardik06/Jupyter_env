````python
import pandas as pd
from sklearn.linear_model import LinearRegression
import pickle

# Dataset
data = {
    'Experience': [1, 2, 3, 4, 5],
    'Salary': [30000, 35000, 40000, 45000, 50000]
}

df = pd.DataFrame(data)

X = df[['Experience']]
y = df['Salary']

# Train model
model = LinearRegression()
model.fit(X, y)

# Save model
with open('model.pkl', 'wb') as f:
    pickle.dump(model, f)

print("Model saved using pickle!")
````

````python
from sklearn.linear_model import LinearRegression
import pandas as pd
import joblib

data = {
    'Experience': [1, 2, 3, 4, 5],
    'Salary': [30000, 35000, 40000, 45000, 50000]
}

df = pd.DataFrame(data)

X = df[['Experience']]
y = df['Salary']

model = LinearRegression()
model.fit(X, y)

# Save model
joblib.dump(model, 'model_joblib.pkl')

print("Model saved using joblib!")
````

# 🚀 Experiment 4: Deployment of ML Model (Using Flask)

## 🔹 What is Model Deployment?

Model deployment means **making your trained ML model available for real-world use** through a web app or API.

👉 Example:

* Enter experience → get predicted salary on a webpage

---

## 🔹 Tools Used

* Flask
* `pickle` / `joblib` (to save model)

---

# 🧠 Concept Flow

```
Train Model → Save Model → Load Model → Create Web App → Take Input → Predict Output
```

---

# 💻 STEP 1: Train & Save Model

### ✔️ Using `pickle`


# 💻 STEP 2: Flask App Code

```python id="flask_app"
from flask import Flask, request, jsonify
import pickle
import numpy as np

# Load model
with open('model.pkl', 'rb') as f:
    model = pickle.load(f)

app = Flask(__name__)

@app.route('/')
def home():
    return "ML Model is Running!"

@app.route('/predict', methods=['POST'])
def predict():
    data = request.get_json()
    experience = np.array([[data['experience']]])
    
    prediction = model.predict(experience)
    
    return jsonify({'Predicted Salary': float(prediction[0])})

if __name__ == '__main__':
    app.run(debug=True)
```

# 📌 Key Concepts Explained

---

## 🔸 1. Pickle

* Used to **serialize (save)** Python objects
* Saves model as `.pkl` file

---

## 🔸 2. Joblib

* Similar to pickle but:

  * Faster for large models
  * Efficient for NumPy arrays

---

## 🔸 3. API (Application Programming Interface)

* Allows communication between systems
* Here: Flask API → sends input → returns prediction

---

## 🔸 4. Flask Routes

* `/` → home page
* `/predict` → prediction endpoint

# ❓ Viva Questions with Answers

---

## ✅ Q1. What is model deployment?

Model deployment is the process of making a trained machine learning model available for real-world use through an application or API.

---

## ✅ Q2. What is Flask?

Flask is a lightweight Python web framework used to build web applications and APIs.

---

## ✅ Q3. What is pickle?

Pickle is a Python module used to serialize and deserialize objects.

---

## ✅ Q4. What is joblib?

Joblib is a library used to efficiently save and load large machine learning models.

---

## ✅ Q5. Difference between pickle and joblib?

| Pickle                | Joblib                  |
| --------------------- | ----------------------- |
| General-purpose       | Optimized for ML        |
| Slower for large data | Faster for large arrays |

---

## ✅ Q6. What is API?

API is a system that allows communication between different software applications.

---

## ✅ Q7. What is endpoint?

An endpoint is a URL where the API receives requests.

---

## ✅ Q8. Why use Flask for ML?

* Easy to use
* Lightweight
* Good for small ML apps

---

## ✅ Q9. What is JSON?

JSON is a format used to send and receive data between client and server.

---

## ✅ Q10. What is request and response?

* Request → input sent to server
* Response → output returned
