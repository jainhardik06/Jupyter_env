````python
# Step 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 2: Create Dataset
data = {
    'Age': [22, 25, 47, 52, 46, 56, 23, 27, 48, 50],
    'Salary': [15000, 29000, 48000, 60000, 52000, 75000, 20000, 32000, 49000, 58000],
    'Purchased': [0, 0, 1, 1, 1, 1, 0, 0, 1, 1]  # Only for evaluation
}

df = pd.DataFrame(data)

X = df[['Age', 'Salary']]
y_true = df['Purchased']  # For metrics comparison

# Step 3: Apply K-Means
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=42)
y_kmeans = kmeans.fit_predict(X)

print("Cluster Labels:", y_kmeans)

# Step 4: Visualization
plt.scatter(X['Age'], X['Salary'], c=y_kmeans)
plt.scatter(kmeans.cluster_centers_[:, 0], 
            kmeans.cluster_centers_[:, 1], 
            s=200, marker='X')

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("K-Means Clustering")
plt.show()

# -------------------------------
# 🔹 Evaluation (Extra - for understanding)
# -------------------------------
from sklearn.metrics import precision_score, recall_score, f1_score

# Adjust labels if needed (since clustering labels can be reversed)
# Try both ways and pick better match
precision = precision_score(y_true, y_kmeans)
recall = recall_score(y_true, y_kmeans)
f1 = f1_score(y_true, y_kmeans)

print("\nPrecision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
````

# 🧪 Experiment 7: K-Means Clustering (with Visualization + Metrics)

---

## 🔹 What is K-Means Clustering?

K-Means is an **unsupervised learning algorithm** used to group data into **K clusters**.

👉 It does NOT use labels (output is unknown)

---

## 🔹 Core Concept

* Choose number of clusters (K)
* Assign data points to nearest centroid
* Update centroids
* Repeat until stable

---

# 🧠 Concepts You Must Know

---

## 🔸 1. Centroid

* Center of a cluster
* Updated in each iteration

---

## 🔸 2. K Value

* Number of clusters
* Must be chosen manually

---

## 🔸 3. Iteration Process

* Assign → Update → Repeat

---

## 🔸 4. Unsupervised Learning

* No labeled output
* Finds hidden patterns

# 📏 Precision, Recall, F1 (Extra Understanding)

Even though K-Means is unsupervised, we compare with actual labels.

---

## 🔹 Precision

* Out of predicted positives, how many are correct

## 🔹 Recall

* Out of actual positives, how many detected

## 🔹 F1 Score

* Balance of precision & recall

# ❓ Viva Questions with Answers

---

## ✅ Q1. What is K-Means?

K-Means is an unsupervised algorithm that groups data into K clusters based on similarity.

---

## ✅ Q2. What is centroid?

Centroid is the center point of a cluster.

---

## ✅ Q3. How does K-Means work?

It assigns points to nearest centroid and updates centroids iteratively until convergence.

---

## ✅ Q4. What is K?

K is the number of clusters.

---

## ✅ Q5. How to choose K?

Using Elbow Method.

---

## ✅ Q6. What is Elbow Method?

A technique to find optimal K by plotting error vs K.

---

## ✅ Q7. What are limitations of K-Means?

* Sensitive to initial centroids
* Requires predefined K
* Not good for non-spherical data

---

## ✅ Q8. Does K-Means need scaling?

✔️ Yes, because it uses distance.

---

## ✅ Q9. What is difference between supervised and unsupervised learning?

| Supervised              | Unsupervised        |
| ----------------------- | ------------------- |
| Has labels              | No labels           |
| Example: Classification | Example: Clustering |

---

## ✅ Q10. Why evaluation is tricky in K-Means?

Because there are no true labels in unsupervised learning.
