````python
# Step 1: Import Libraries
import pandas as pd
import numpy as np

# Step 2: Create Dataset
data = {
    'Age': [22, 25, 47, 52, 46, 56, 23, 27, 48, 50],
    'Salary': [15000, 29000, 48000, 60000, 52000, 75000, 20000, 32000, 49000, 58000],
    'Experience': [1, 2, 5, 7, 6, 8, 2, 3, 6, 7]
}

df = pd.DataFrame(data)

X = df.values

# Step 3: Feature Scaling (IMPORTANT)
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_scaled = sc.fit_transform(X)

# Step 4: Apply PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)  # reduce to 2 features
X_pca = pca.fit_transform(X_scaled)

print("Original Shape:", X.shape)
print("Reduced Shape:", X_pca.shape)

print("\nPCA Data:\n", X_pca)
````

````python
import matplotlib.pyplot as plt

plt.scatter(X_pca[:, 0], X_pca[:, 1])
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization")
plt.show()
````

# 🧪 Experiment 10: PCA (Principal Component Analysis – Feature Selection)

---

## 🔹 What is PCA?

PCA is an **unsupervised dimensionality reduction technique** used to:

* Reduce number of features
* Preserve maximum important information (variance)

👉 Converts original features → **new features (principal components)**

---

## 🔹 Why PCA is Used?

* Reduce complexity
* Remove redundant features
* Improve performance
* Visualization of high-dimensional data

---

# 🧠 Core Concept

👉 PCA finds directions (components) where data varies the most.

* PC1 → maximum variance
* PC2 → second maximum variance

---

# 💻 Basic Code (Simple + Exam Ready)
