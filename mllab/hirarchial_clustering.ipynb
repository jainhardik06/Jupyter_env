```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering

data = {
    "Age": [22, 25, 47, 52, 46, 56, 23, 27, 48, 50],
    "Salary": [15000, 29000, 48000, 60000, 52000, 75000, 20000, 32000, 49000, 58000],
    "Experience": [1, 2, 5, 7, 6, 8, 2, 3, 6, 7],
    "Score": [65, 70, 88, 90, 85, 92, 68, 72, 89, 91]
}

df = pd.DataFrame(data)
X = df[["Age", "Salary", "Experience", "Score"]]

hc = AgglomerativeClustering(n_clusters=2)
labels = hc.fit_predict(X)

print("Cluster Labels:", labels)
```

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X, method='ward')

dendrogram(linked)
plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.show()
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1, 6):
    km = KMeans(n_clusters=i, random_state=42)
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1, 6), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X, labels)
print("Silhouette Score:", score)
```

```python
from sklearn.datasets import make_blobs

X_real, _ = make_blobs(n_samples=100, centers=3, random_state=42)

hc_real = AgglomerativeClustering(n_clusters=3)
labels_real = hc_real.fit_predict(X_real)

plt.scatter(X_real[:, 0], X_real[:, 1], c=labels_real)
plt.title("Hierarchical Clustering (Real Data)")
plt.show()
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X2 = pca.fit_transform(X)

plt.scatter(X2[:, 0], X2[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Hierarchical Clustering (4 features, projected to 2D)")
plt.show()
```

## 🔹 What is Hierarchical Clustering?

Hierarchical Clustering is an **unsupervised learning algorithm** that builds a **tree-like structure (dendrogram)** of clusters.

👉 No need to predefine K (unlike K-Means)

---

## 🔹 Types

* **Agglomerative** (bottom → up) ✅ mostly used
* **Divisive** (top → down)

---

## 🔹 Core Idea

* Start with each point as a cluster
* Merge closest clusters step by step
* Continue until one cluster remains
