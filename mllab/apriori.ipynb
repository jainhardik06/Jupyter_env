````python
# Step 1: Import Libraries
import pandas as pd
import numpy as np

# Step 2: Create Dataset (Transactions)
transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter']
]

# Convert to DataFrame
from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)

df = pd.DataFrame(te_data, columns=te.columns_)

print("Dataset:\n", df)

# Step 3: Apply Apriori
from mlxtend.frequent_patterns import apriori, association_rules

frequent_items = apriori(df, min_support=0.4, use_colnames=True)

print("\nFrequent Itemsets:\n", frequent_items)

# Step 4: Generate Rules
rules = association_rules(frequent_items, metric="confidence", min_threshold=0.6)

print("\nAssociation Rules:\n", rules)
````

# 🧪 Experiment 9: Apriori Algorithm

---

## 🔹 What is Apriori Algorithm?

Apriori is an **unsupervised learning algorithm** used for **association rule mining**.

👉 It finds:

* Items that occur **together frequently**

---

## 🔹 Real-Life Example

In a supermarket:

* Customers who buy **bread** also buy **butter**

👉 Rule:

```
Bread → Butter
```

---

# 🧠 Core Concepts

---

## 🔸 1. Support

* How frequently item appears in dataset

[
Support = \frac{Transactions\ containing\ item}{Total\ Transactions}
]

---

## 🔸 2. Confidence

* Probability that item B is bought when A is bought

[
Confidence = \frac{Support(A \cap B)}{Support(A)}
]

---

## 🔸 3. Lift

* Strength of rule

[
Lift = \frac{Confidence}{Support(B)}
]

👉 Lift > 1 → strong rule

# 🧠 Important Concept (Apriori Property)

👉 “If an itemset is frequent, then all its subsets are also frequent”

This helps reduce computation.

# ❓ Viva Questions with Answers

---

## ✅ Q1. What is Apriori Algorithm?

Apriori is an unsupervised algorithm used to find frequent itemsets and generate association rules from transactional data.

---

## ✅ Q2. What is support?

Support is the frequency of occurrence of an itemset in the dataset.

---

## ✅ Q3. What is confidence?

Confidence is the probability that item B is purchased when item A is purchased.

---

## ✅ Q4. What is lift?

Lift measures the strength of an association rule.
Lift > 1 indicates a strong relationship.

---

## ✅ Q5. What is association rule?

A rule that shows relationship between items (e.g., A → B).

---

## ✅ Q6. What is Apriori property?

If a set is frequent, all its subsets must also be frequent.

---

## ✅ Q7. Applications of Apriori?

* Market basket analysis
* Recommendation systems
* Retail analysis

---

## ✅ Q8. What are limitations?

* High computation
* Slow for large datasets

---

## ✅ Q9. Difference between classification and association?

| Classification | Association        |
| -------------- | ------------------ |
| Predict label  | Find relationships |
| Supervised     | Unsupervised       |

---

## ✅ Q10. What is transactional data?

Data that contains list of items bought together in a transaction.
