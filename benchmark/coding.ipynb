```python
# Segregating Even and Odd Product IDs
# ○ Problem: You are given an array of integers representing product IDs. Write a program to segregate the even IDs to the left side of the array and the odd IDs to the right side.
# ○ Input: [12, 34, 45, 9, 8, 90, 3]
# ○ Output: [12, 34, 8, 90, 45, 9, 3] (Note: the exact order of evens and odds does not matter, as long as they are segregated).

inp = [12, 34, 45, 9, 8, 90, 3]
temp = []
for i in inp:
    if i%2 == 0:
        temp.append(i)

for i in inp:
    if i in temp:
        continue
    else:
        temp.append(i)
print(temp)
```

```python
# Q3. Summing Employee Salaries within a Specific Range
# ○ Problem: You are given an array of employee salaries and two integers, low and high. Return the sum of all salaries that fall strictly within the inclusive range [low, high].
# ○ Input: salaries = [10, 5, 15, 20, 25, 30], low = 15, high = 25
# ○ Output: 60 (15 + 20 + 25)
salaries = [10, 5, 15, 20, 25, 30] 
low = 15 
high = 25
sum = 0
for i in salaries:
    if low <= i <= high:
        sum+=i
print(sum) 
```

```python
# Q4. Present in One Absent in Another
# ○ Problem: Given two arrays, find all elements that are present in the first array but completely absent from the second array.
# ○ Input: A = [1, 2, 3, 4, 5], B = [2, 4, 6]
# ○ Output: [1, 3, 5]
A = [1, 2, 3, 4, 5] 
B = [2, 4, 6]
t =[]
for i in A:
    if i in B:
        continue
    else:
        t.append(i)
print(t)
```

```python
# Q5. Minimum Removals to Eliminate 'ppp' Substrings
# ● Problem: Given a string consisting of lowercase letters, find the minimum number of characters you need to remove so that the string does not contain three consecutive 'p's ("ppp") anywhere.
# ● Input: "bpppaqpppp"
# ● Output: 3 (Remove one 'p' from the first group and one 'p' from the second group to get "bppaqppp", wait, remove another to get "bppaqpp")

b = "bpppaqpppp"
a = list(b)
i=0
c=0
while(i <= len(a)-1):
    print(a[i])
    if a[i:i+3] == ['p','p','p']:
        a.pop(i)
        c+=1
        i-=1
    else:
        i+=1
print(a,c)
```

```python
# Q8. Count Unique Characters in a String
# ● Problem: Given a string, write a program to count the exact number of unique, non-repeating characters present in it (characters that appear exactly once).
# ● Input: "programming"
# ● Output: 5 (The characters 'p', 'o', 'a', 'i', 'n' appear exactly once)

a = "programming"
l = []
for i in a:
    t = a.count(i)
    if t == 1:
        l.append(i)
print(len(l))
    
```

```python
# Q7. Digital Grouping with Custom Separator
# ● Problem: You are given a long string of digits and a character separator. Format the string so that digits are grouped in blocks of 3, separated by the character. If the last block only has 1 digit left, reformat the end to be two blocks of 2.
# ● Input: digits = "12345678", separator = "-"
# ● Output: "123-456-78"

a = "1234567890123456"
for i in range(0,len(a),3):
    if i+4 == len(a):
        print(a[i:i+2],end="-")
        print(a[i+2:], end="")
        break
    elif i+3 >= len(a):
        print(a[i:i+3], end="")
    else:
        print(a[i:i+3], end="-")
```
