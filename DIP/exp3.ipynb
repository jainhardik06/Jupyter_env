```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt
import cv2 as cv
import urllib.request
```

```python

url1 = "https://vrindavantoday.in/wp-content/uploads/2023/09/309210901_10162146713854838_3487854663768239969_n.jpg"
url2 = "https://temple.yatradham.org/public/Product/temple/temple_98qxh8GE_202408041807060.jpg"


def load_image_from_url(url):
    req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
    response = urllib.request.urlopen(req)
    arr = np.asarray(bytearray(response.read()), dtype=np.uint8)
    img = cv.imdecode(arr, cv.IMREAD_COLOR)
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    img_resized = cv.resize(img_rgb, (500, 500))
    return img_resized


img1 = load_image_from_url(url1)
img2 = load_image_from_url(url2)


plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(img1)
plt.title('Image 1')
plt.axis('off')

plt.subplot(122)
plt.imshow(img2)
plt.title('Image 2')
plt.axis('off')

plt.show()
```

```python
h1,w1,c1 = img1.shape
h2,w2,c2 = img2.shape
print("Height and Width of Image 1:", h1, w1)
print("Height and Width of Image 2:", h2, w2)

# 
```

```python
add = np.clip(img1 + img2, 0, 255).astype(np.uint8)
sub = np.clip(img1 - img2, 0, 255).astype(np.uint8)
mul = np.clip(img1 * 1.5, 0, 255).astype(np.uint8)
div = np.clip(img1 / 2, 0, 255).astype(np.uint8)

def show_image_operations(images, titles):
    plt.figure(figsize=(16, 8))
    for i in range(len(images)):
        plt.subplot(2, 2, i + 1)
        plt.imshow(images[i])
        plt.title(titles[i])
        plt.axis('off')
    plt.show()

show_image_operations(
    [add, sub, mul, div],
    ['Addition', 'Subtraction', 'Multiplication', 'Division']
)
```
