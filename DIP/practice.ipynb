1. download an image in jpg or jpeg format, 
    a. Find its Height and Width
    b. BGG plotting
    c. RGB plotting
    d. RGB to Grey
    e. Mirror the image w.r.t --> x,y,xy
    f. Flip the image w.r.t --> x,y,xy

2. Print indexed image(//64)

```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt
import cv2 as cv
import os
from skimage import io
from PIL import Image
```

```python
img = cv.imread('pisa.jpg')
height, width, channels = img.shape
print("Height:", height, "Width:", width)
```

```python
plt.figure(figsize=(15,5))
plt.subplot(131)
plt.title('BGR Plotting')
plt.imshow(img)
plt.axis('off')

img_rgb=cv.cvtColor(img,cv.COLOR_BGR2RGB)
plt.subplot(132)
plt.title('RGB Plotting')
plt.imshow(img_rgb)
plt.axis('off')

img_gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
plt.subplot(133)
plt.title('Grayscale Image')
plt.imshow(img_gray, cmap='gray')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(18,6))
plt.subplot(141)
plt.title('Original Image')
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(142)
plt.title('Mirror Image on x axis')
plt.imshow(img_rgb[:,::-1])
plt.axis('off')

plt.subplot(143)
plt.title('Mirror Image on y axis')
plt.imshow(img_rgb[::-1,:])
plt.axis('off')

plt.subplot(144)
plt.title('Mirror Image on x and y axis')
plt.imshow(img_rgb[::-1,::-1])
plt.axis('off')
plt.show()
```

```python
img_fx = cv.flip(img_rgb, 0)
img_fy = cv.flip(img_rgb, 1)
img_fxy = cv.flip(img_rgb, -1)

plt.figure(figsize=(18,6))
plt.subplot(131)
plt.title('Flip around x axis')
plt.imshow(img_fx)
plt.axis('off')

plt.subplot(132)
plt.title('Flip around y axis')
plt.imshow(img_fy)
plt.axis('off')

plt.subplot(133)
plt.title('Flip around x and y axis')
plt.imshow(img_fxy)
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
indexed = img_rgb//64
indexed_display = indexed * 64

plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.title('Original Image')
plt.imshow(img_rgb)
plt.axis('off')

plt.subplot(122)
plt.title('Indexed Image')
plt.imshow(indexed_display.astype(np.uint8))
plt.axis('off')
plt.tight_layout()
plt.show()
```
