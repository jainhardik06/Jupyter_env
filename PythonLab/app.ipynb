```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def hello():
    return 'hello world'

if __name__ == '__main__':
    app.run()
```

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def home():
    return "This is Home Page"

@app.route('/about')
def about():
    return "This is About Page"

if __name__ == '__main__':
    app.run()
```

```python
from flask import Flask, url_for

app = Flask(__name__)

@app.route('/')
def home():
    return "Go to About Page: " + url_for('about')

@app.route('/about')
def about():
    return "Welcome to About Page"

if __name__ == '__main__':
    app.run()
```

```python
from flask import Flask

app = Flask(__name__)

@app.route('/user/<name>')
def user(name):
    return f"Hello {name}"

if __name__ == '__main__':
    app.run()
```

```python
from flask import Flask, request

app = Flask(__name__)

@app.route('/', methods=['GET', 'POST'])
def form():
    if request.method == 'POST':
        name = request.form.get('name')
        return f"Hello {name}"
    
    return '''
        <form method="post">
            Enter Name: <input type="text" name="name">
            <input type="submit">
        </form>
    '''

if __name__ == '__main__':
    app.run()
```
