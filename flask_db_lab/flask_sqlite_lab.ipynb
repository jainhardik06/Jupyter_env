# Flask + SQLite Lab\n
Create table, insert student data, and display records.

```python
import sqlite3\n
\n
conn = sqlite3.connect('students.db')\n
conn.execute('''\n
CREATE TABLE IF NOT EXISTS students (\n
    id INTEGER PRIMARY KEY AUTOINCREMENT,\n
    name TEXT,\n
    age INTEGER,\n
    course TEXT\n
)\n
''')\n
conn.close()\n
print('Table Created Successfully')
```

```python
from flask import Flask, render_template, request, redirect, url_for\n
import sqlite3\n
\n
app = Flask(__name__)\n
\n
@app.route('/')\n
def home():\n
    return render_template('form.html')\n
\n
@app.route('/add', methods=['POST'])\n
def add():\n
    name = request.form['name']\n
    age = request.form['age']\n
    course = request.form['course']\n
\n
    conn = sqlite3.connect('students.db')\n
    conn.execute('INSERT INTO students (name, age, course) VALUES (?, ?, ?)', (name, age, course))\n
    conn.commit()\n
    conn.close()\n
\n
    return redirect(url_for('students'))\n
\n
@app.route('/students')\n
def students():\n
    conn = sqlite3.connect('students.db')\n
    data = conn.execute('SELECT * FROM students').fetchall()\n
    conn.close()\n
    return render_template('display.html', data=data)\n
\n
app.run()
```
